# Projeto Final do Aprofundamento de Analytics - E-commerce

### Projeto de Dashboard de E-commerce

Este projeto visa aplicarem habilidades essenciais de análise de dados em um cenário prático e realista. Vamos trabalhar com um conjunto de dados de transações de clientes de uma loja virtual, distribuídos em duas tabelas distintas. O objetivo final é construir um dashboard interativo que facilite a visualização e análise das informações relevantes do e-commerce, utilizando ferramentas Looker Studio.

### Objetivo do Projeto:

Tratamento de Dados: Realizar a junção (JOIN) de duas tabelas utilizando SQL para consolidar as informações.
Análise de Dados: Exportar os dados resultantes para um arquivo CSV.
Visualização de Dados: Desenvolver um dashboard interativo e informativo para visualização das principais métricas e insights do e-commerce.

### Tabelas Disponibilizadas:

**Tabela de Transações:** Contém os registros de transações realizadas pelos clientes, incluindo detalhes como ID da transação, valor e outros.


**Tabela de Dados Pessoais:** Contém as informações pessoais dos clientes, como ID do cliente, nome, genero, cidade, etc.

**Chave de Ligação:** As tabelas se relacionam através da coluna ID_CLIENT, que é a chave identificadora dos clientes.

### Etapas do Projeto:

1. Realizar jução das duas tabelas, unificando as informações através da coluna ID_CLIENT.

2. Exportar os dados consolidados resultantes do JOIN para um arquivo CSV.

3. Utilizar Looker Studio para importar o arquivo CSV.

4. Criar visualizações interativas que apresentem métricas importantes, como total de vendas, número de transações, distribuição geográfica dos clientes, perfil demográfico dos clientes, entre outros.

Abaixo temos a configuração do ambiente SQL:

```python
# Importando as bibliotecas necessárias

import sqlite3
import pandas as pd
```

```python
# Carregando as bases de dados em dataframes

df_transacoes = pd.read_csv(r"Bases\TB_TRANSACOES_PROJETO_ECOMM.csv", delimiter=';')
df_clientes = pd.read_csv(r"Bases\TB_CLIENTES_PROJETO_ECOMM.csv", delimiter=';')
```

```python
# Conectando ao banco de dados SQLite (ou criando-o se não existir)

conn = sqlite3.connect('projeto.db')
# Carregar o DataFrame no banco de dados SQLite - criando tb_transacoes e tb_clientes
df_transacoes.to_sql('tb_transacoes', conn, index=False, if_exists='replace')
df_clientes.to_sql('tb_clientes', conn, index=False, if_exists='replace')
```

```python
# Função para executar consultas SQL e retornar o resultado como um DataFrame
def run_query(query):
    return pd.read_sql_query(query, conn)
```

```python
# Consulta SQL para unir as tabelas e tratar valores nulos

query = """
SELECT tr.id_client,
       IFNULL(cl.first_name, 'N/Name') AS First_Name, 
       IFNULL(cl.gender, 'N/Gender') AS Gender, 
       IFNULL(cl.job_title, 'N/Job title') AS Job_Title,
       IFNULL(cl.state_name, 'N/State') AS State_Name,
       tr.category,
       tr.card_type,
       ( Cast(Replace(tr.price, ',', '.') AS FLOAT) ) AS Price
FROM   tb_clientes AS cl
       RIGHT JOIN tb_transacoes AS tr
               ON cl.id_client = tr.id_client;
"""
result_df = run_query(query)
print(result_df)
```

Justificativa para o Uso de `RIGHT JOIN`

Para uma análise abrangente das vendas de uma loja virtual, o uso de um `RIGHT JOIN` entre as tabelas de clientes e transações é a abordagem mais adequada.

Ao aplicar o `RIGHT JOIN` da tabela de clientes (TB_CLIENTES_PROJETO_ECOMM) com a tabela de transações (TB_TRANSACOES_PROJETO_ECOMM), garantimos que **todas as vendas sejam incluídas na análise**. Essa abordagem prioriza a tabela de transações, que é o ponto central da avaliação, e anexa os dados dos clientes correspondentes a cada venda.

O principal benefício dessa estratégia é a possibilidade de identificar e agir sobre as transações que não têm um cliente formalmente registrado. Isso não apenas permite uma análise completa de todas as vendas, mas também destaca inconsistências nos dados, permitindo que a empresa tome medidas para corrigir e complementar as informações de clientes ausentes.

Exportando o arquivo como CSV:

```python
result_df.to_csv('dados_ecommerce_final.csv', index=False)
```
